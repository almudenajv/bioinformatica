#### 4.- Dada la lista de tuplas _pares_. Filtra solo las tuplas donde el primer número es mayor que el segundo. Devuelve a una lista con True o False indicando si la suma de cada tupla es par.

```python
pares = [(1,5),(2,2),(10,3),(4,4)]
```

```python
pares_filtrados = [t for t in pares if t[0] > t[1]]

resultado = [(t[0] + t[1]) % 2 == 0 for t in pares_filtrados]

print("Tuplas filtradas:", pares_filtrados)
print("¿La suma es par?:", resultado)
```

#### 8.- Dado el diccionario _usuarios_, convierte todos los nombres a mayúsculas. Crea un set con todos los roles del sistema. Devuelve un diccionario donde cada rol sea clave y su valor, la lista de usuarios que tienen ese rol.

```python
usuarios = {
    "u1" : {"nombre":"ana","roles":{"admin","user"}},
    "u2" : {"nombre":"luis","roles":{"user"}},
    "u3" : {"nombre":"marta","roles":{"user","moderator"}}
}
```

```python
roles_sistema = set()
usuarios_por_rol = {}

for uid, info in usuarios.items():
    info["nombre"] = info["nombre"].upper()
    nombre_actual = info["nombre"]
    for rol in info["roles"]:
        roles_sistema.add(rol)
        if rol not in usuarios_por_rol:
            usuarios_por_rol[rol] = []
        usuarios_por_rol[rol].append(nombre_actual)

print("Diccionario de usuarios actualizado:")
print(usuarios)
print("\nSet de roles del sistema:")
print(roles_sistema)
print("\nDiccionario de usuarios por rol:")
print(usuarios_por_rol)
```

#### 9.- Dada la lista _valores_, convierte cada número a string, pero con las siguientes reglas:
- Si es par: "par:<num>"
- Si es múltiplo de 4: "mult3:<num>"
- Si cumple ambas condiciones, concatenar: "par-mult3:<num>"
#### Devuelve la lista final.

```python
valores = [10,21,33,42,55]
```

```python
resultado = []

for n in valores:
    es_par = (n % 2 == 0)
    es_mult3 = (n % 3 == 0)
    
    if es_par and es_mult3:
        resultado.append(f"par-mult3:{n}")
    elif es_par:
        resultado.append(f"par:{n}")
    elif es_mult3:
        resultado.append(f"mult3:{n}")
    else:
        resultado.append(str(n))  # Si no cumple ninguna, se deja como string normal

print(resultado)
```

#### 10.- Dado el dicionario _datos_, pon la primera letra de "nombre" en mayúsculas. Calcula la suma total de "puntos" y guarda el total en una nueva clave "total". Marca un booleano como "aprobado" si el total es mayor a 50. Finalmente, crea un set con todos los nombres de los usuarios aprobados. 

```python
datos = [
    {"id":1,"nombre":"ana","puntos":(10,20,30)},
    {"id":2,"nombre":"luis","puntos":(5,15)},
    {"id":3,"nombre":"marta","puntos":(100,50,25)}
]
```

```python
aprobados = set()

for usuario in datos:
    usuario["nombre"] = usuario["nombre"].capitalize()
    total_puntos = sum(usuario["puntos"])
    usuario["total"] = total_puntos
    if total_puntos > 50:
        usuario["aprobado"] = True
        aprobados.add(usuario["nombre"])
    else:
        usuario["aprobado"] = False

print("Lista de datos actualizada:")
print(datos)
print("\nSet de usuarios aprobados:")
print(aprobados)
```

#### 12.- Dado el diccionario _estudiantes_, crea un nuevo diccionario con la siguiente estructura:
- La clave debe ser el nombre en mayúsculas
- El valor debe ser otro diccionario con:
    - "promedio" como promedio de notas
    - "aprobado" como True si el promedio es >= 20, de lo contrario es False
    - "max_min" como tupla con (nota_max,nota_min)

```python
estudiantes = {
    "s1" : {"nombre":"ana","notas":[10,20,30]},
    "s2" : {"nombre":"luis","notas":[15,5]},
    "s3" : {"nombre":"marta","notas":[50,100,25]},
    "s4" : {"nombre":"pepe","notas":[40]}
}
```

```python
nuevo_diccionario = {}

for id_estudiante, info in estudiantes.items():
    nombre_mayus = info["nombre"].upper()
    notas = info["notas"]
    promedio = sum(notas) / len(notas)
    aprobado = promedio >= 20
    max_min = (max(notas), min(notas))
    nuevo_diccionario[nombre_mayus] = {
        "promedio": promedio,
        "aprobado": aprobado,
        "max_min": max_min
    }

print(nuevo_diccionario)
```

#### 13.- Crea un archivo llamado _notas_alumnos.txt_ con este contenido:
#### Ana,10,20,30
#### Luis,5,15
#### Marta,100,50,25
#### Convierte cada línea en un diccionario con "nombre" capitalizado, "promedio" y "aprobado" si promedio >= 20.
#### Guarda la lista final en un archivo resultado.json

```python
import json
archivo = open("notas_alumnos.txt", "w")
archivo.write("Ana,10,20,30\n")
archivo.write("Luis,5,15\n")
archivo.write("Marta,100,50,25\n")
archivo.close()

lista_resultado = []
archivo = open("notas_alumnos.txt", "r")

for linea in archivo:
    linea = linea.strip()
    if linea:
        partes = linea.split(",")
        nombre = partes[0].capitalize()
        notas = [float(n) for n in partes[1:]]
        
        promedio = sum(notas) / len(notas) if notas else 0.0
        aprobado = promedio >= 20
        
        dic_alumno = {
            "nombre": nombre,
            "promedio": promedio,
            "aprobado": aprobado
        }
        lista_resultado.append(dic_alumno)

archivo.close()

archivo_json = open("resultado.json", "w")
json.dump(lista_resultado, archivo_json, ensure_ascii=False, indent=4)
archivo_json.close()
```

#### 14.- Crea el archivo _usuarios.json_con el contenido del diccionario _datos_. Abre el archivo de nuevo y convierte todos los nombres a mayúsculas, luego crea un archivo _usuarios.csv_ con formato:
- id,nombre,roles
- 1,ANA,user|admin
- 2,LUIS,user
- 3,MARTA,moderator|user

```python
import json
import csv

datos = [
    {"id": 1, "nombre": "ana", "roles": ["admin", "user"]},
    {"id": 2, "nombre": "luis", "roles": ["user"]},
    {"id": 3, "nombre": "marta", "roles": ["user", "moderator"]}
]

archivo_json = open("usuarios.json", "w", encoding="utf-8")
json.dump(datos, archivo_json, ensure_ascii=False, indent=4)
archivo_json.close()

archivo_json_lectura = open("usuarios.json", "r", encoding="utf-8")
datos_cargados = json.load(archivo_json_lectura)
archivo_json_lectura.close()

for item in datos_cargados:
    item["nombre"] = item["nombre"].upper()

archivo_csv = open("usuarios.csv", "w", encoding="utf-8", newline="")
escritor = csv.writer(archivo_csv)

escritor.writerow(["id", "nombre", "roles"])

for item in datos_cargados:
    roles_str = "|".join(item["roles"])
    escritor.writerow([item["id"], item["nombre"], roles_str])

archivo_csv.close()
```

```python
import re
import json

archivo = open("fragmento-libro.txt", "w", encoding="utf-8")
archivo.write("En un lugar de la Mancha, de cuyo nombre no quiero acordarme... [texto del fragmento] ...\n")
archivo.close()


archivo_lectura = open("fragmento-libro.txt", "r", encoding="utf-8")
texto = archivo_lectura.read()
archivo_lectura.close()


palabras = re.findall(r'\b\w+\b', texto.lower())

frecuencias = {}
for palabra in palabras:
    frecuencias[palabra] = frecuencias.get(palabra, 0) + 1

mas_comunes = sorted(frecuencias.items(), key=lambda x: x[1], reverse=True)[:5]
diccionario_top5 = {palabra: freq for palabra, freq in mas_comunes}

archivo_json = open("frecuencias.json", "w", encoding="utf-8")
json.dump(diccionario_top5, archivo_json, ensure_ascii=False, indent=4)
archivo_json.close()
```

#### 17.- Crea un archivo _alumnos.json_ con el contenido siguiente:
- [{"nombre": "ana", "notas": [10, 20, 30]},{"nombre": "luis", "notas": [5, 15]},{"nombre": "marta", "notas": [100, 50, 25]}]
#### Lee el archivo y genera un diccionario donde la clave sea el nombre y el valor tenga el siguiente formato:
- {"promedio":X,"max":Y,"min":Z,"aprobado":True/False}


```python
import json

datos = [
    {"nombre": "ana", "notas": [10, 20, 30]},
    {"nombre": "luis", "notas": [5, 15]},
    {"nombre": "marta", "notas": [100, 50, 25]}
]

archivo = open("alumnos.json", "w", encoding="utf-8")
json.dump(datos, archivo, ensure_ascii=False, indent=4)
archivo.close()

archivo_lectura = open("alumnos.json", "r", encoding="utf-8")
alumnos_cargados = json.load(archivo_lectura)
archivo_lectura.close()

resultado = {}

for alumno in alumnos_cargados:
    nombre = alumno["nombre"]
    notas = alumno["notas"]
    
    promedio = sum(notas) / len(notas) if notas else 0
    max_nota = max(notas) if notas else 0
    min_nota = min(notas) if notas else 0
    aprobado = promedio >= 20  
    
    resultado[nombre] = {
        "promedio": promedio,
        "max": max_nota,
        "min": min_nota,
        "aprobado": aprobado
    }

print(json.dumps(resultado, ensure_ascii=False, indent=4))
```
