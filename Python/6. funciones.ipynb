```python
def es_par(lista_numeros):
    lista_pares = []
    for numero in lista_numeros:
        if numero % 2 == 0:
            lista_pares.append(numero)
        else:
            pass
    return lista_pares
```

```python
es_par([1,2,3,4,5,6])
```

```python
es_par([2,4,6])
```

```python
es_par([1,3,5])
```

```python
from random import shuffle
mi_lista = [' ','O',' ']
def desordenador_lista(lista):
    shuffle(lista)
    return lista

desordenar_lista(mi_lista)
mi_lista
```

```python
def adivinacion():
    intento = ''
    while intento not in ['0','1','2']:
        intento = input("Escoje un numero: 0, 1, 2: ")
    return int(intento)
```

```python
adivinacion()
```

```python
def comprobacion(lista,intento):
    if lista[intento] == 'O':
        print('has acertado')
    else: 
        print('has fallado, suerte para la proxima')
        print(lista)
```

```python
comprobacion(mi_lista,intento)

mi_lista = [' ','O',' ']
lista_desordenada=desordenar_lista(mi_lista)
intento = adivinacion()
comprobacion(lista_desordenada,intento)
```
