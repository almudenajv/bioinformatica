# WHILE LOOPS

```python
x = 0
while x < 10:
    print('x is current: ',x)
    print(' x is still less than 10, adding 1 to x')
    x+=1
```

```python
x = 0
while x < 10:
    print('x is current: ',x)
    print(' x is still less than 10, adding 1 to x')
    x+=1
else:
    print('All done!')
```

```python
x = 0
while x < 10:
    print('x is current: ',x)
    print(' x is still less than 10, adding 1 to x')
    x+=1
    if x == 3:
        print('x==3')
    else:
        print('continuing…!')
        continue
```

```python
x = 0
while x < 10:
    print('x is current: ',x)
    print(' x is still less than 10, adding 1 to x')
    x+=1
    if x == 3:
        print('breaking because x==3')
        break
    else:
        print('continuing…!')
        continue
```

```python
lista_prueba = ['huevo','pan','leche','azucar','cafe','nutella']
count = 0
for i in lista_prueba:
    print(i)
    count += 1
    print(count)

print (f'ese es el valor final del contador [count]')

condicion = True
count2 = 0
while condicion:
    lista_prueba.append('tomate')
    print(lista_prueba)
    count2 += 1 
    if count2 == 7:
        condicion=False
```
